```python
def is_odd(number):
    return number % 2 == 1

def is_even(number):
    return number % 2 == 0

def is_prime(number):
    if number == 0 or number == 1:
        return False
    for i in range(2, number-1):
        if number % i == 0:
            return False
    return True

def is_square(number):
    i = 0
    sq = 0
    while sq < number:
        i+=1
        sq = i*i

    return sq == number


def is_cube(number):
    i = 0
    cube = 0
    while cube < number:
        i+=1
        cube = i*i*i

    return cube == number

def input_integer():
    while True:
        data = input("Give an integer: ").strip()
        if data.isdigit():
            return int(data)
        else:
            print("Wrong, only digits please.")

def input_float():
    while True:
        response = input("Give a float: ").strip()
        if "." in response:
            parts = response.split(".")
            if len(parts) > 2:
                print("Only one dot at most.")
            elif parts[0].isdigit() and parts[1].isdigit():
                return float(response)
            else:
                print("Only digits please.")
        else:
            if response.isdigit():
                return float(response)
            else:
                print("Digits only please!")
```
