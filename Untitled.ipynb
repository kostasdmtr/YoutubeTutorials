```python
#globals
students = [
    {
        "id":1001,
        "name":"James",
        "surname":"Hetfield",
        "father's name":"John",
        "age":15,
        "class":1,
        "card_number":"AK321657",
    },
    {
        "id":1002,
        "name":"Lars",
        "surname":"Ulrich",
        "father's name":"Michael",
        "age":16,
        "class":2,
        "card_number":"AK123987",
    },
    {
        "id":1003,
        "name":"Rob",
        "surname":"Truillo",
        "father's name":"Yeabuddy",
        "age":17,
        "class":3,
        "card_number":"AK682054",
    }
]

def print_pupil(pupil):
    if pupil in students:
        for key, value in pupil.items():
            print(f"{key} -> {value}")

def create_pupil():
    completely_new = True
    temp_name = str(input("Student's name: "))
    temp_surname = str(input("Student's surname: "))
    temp_fathers_name = str(input("Father's name: "))
    for student in students:
        if(student["name"] == temp_name and student["surname"] == temp_surname and student["father's name"] == temp_fathers_name):
            completely_new = False
            print("\n----- OPTION 1 ERROR: THIS STUDENT ALREADY EXISTS. -----\n")
            print("Its elements are: ")
            for key,value in student.items():
                print(str(key)+"-> "+str(value))
            print("\nDo you still wanna push the other elements of his?")
            yesorno = str(input("Yes(y) or No(n)?: "))
            if yesorno == "y":
                student['id'] = max(student['id']+1 for student in students)
                print(student['id'])
                student["name"] = temp_name
                student["surname"] = temp_surname
                student["father's name"] == temp_fathers_name
                student['age'] = int(input("Student's age: "))
                student['class'] = int(input("Student's class: "))
                student['card_number'] = str(input("Student's card number")) 
            else:
                return None

    if completely_new:
        student['id'] = next_id()
        print(f"Student's ID: {student['id']}")
        student["name"] = temp_name
        student["surname"] = temp_surname
        student["father's name"] = temp_fathers_name
        student['age'] = int(input("Student's age: "))
        student['class'] = int(input("Student's class: "))
        student['card_number'] = str(input("Student's card number"))
    print("------ |CONGRATS| ----- |NEW RECORD JUST SAVED| -----")
    
    print_pupil(students[len(students)-1])
    return students[len(students)-1]

def next_id():
    return max(student["id"] for student in students) +1



def main():
    
    while True:
    
        print("----- MENU -----")
        print("Option 1: New record.")
        print("Option 2: Print.")
        print("Option 3: Update existing record.")
        print("Option 4: Delete existing record.")
        print("Option 5: Exit")
    
        answer = input("Give your answer(Options 1-5): ")
        if (not answer.isnumeric() or int(answer) < 1 or int(answer) > 5):
            print("\n !ERROR! \nNot appropriate option number, try again!\n")
            continue
    
        #option 1
        if (int(answer) == 1):
            pupil = create_pupil()
            if pupil is None:
                continue
            else:
                print("NEW PUPIL")
                print_pupil(pupil)
                
                
    
        #option 2
        if (int(answer) == 2):
            temp_name = str(input("For which student you want info?: "))
            found = False
            for student in students:
                if temp_name == student["name"]:
                    found = True
                    found_student = student
                    
            if found:
                print("\nStudent found, printing his elements...")
                print_pupil(found_student)
            else:
                print("Student with these surname doesn't exist.")
    
                
        #option 3
        if (int(answer) == 3):
            found = False
            id_to_update = int(input("Give me the ID of the student you want to update its elements: "))
            for student in students:
                if student['id'] == id_to_update:
                    student_to_update = student
                    found = True
    
            if (found):
                print("His current elements are...:")
                print_pupil(student_to_update)
                #na kanw to update twra
                attribute = str(input("Give me exactly which element you want to modify: "))
                if attribute in student_to_update.keys():
                    student_to_update[attribute] = str(input(f"Give his {attribute} new value: "))
                else:
                    print("To attribute den yparxei")
            else:
                print("Student doesn't exist!")
                continue
    
        #option 4 (delete)
        if (int(answer) == 4):
            to_delete_id = int(input("Which student you would like to delete from the database based on his ID?"))
            found_to_delete = False
            position = -1
            for student in students:
                position += 1
                if to_delete_id == student["id"]:
                    found_to_delete = True
                    student_to_delete = student
                    break
        
            if found_to_delete:
                print("Found him and now I am deleting him.")
                students.pop(position)
            else:
                print("Student doesn't exist.")
        
            print(students)
            print(len(students))
            for student in students:
                print(student["id"])
        
        #option 5
        if (int(answer) == 5):
            leave = str(input("Sure you want to exit? (y/n)"))
            if (leave == "y"):
                break
            else:
                continue
    
        
        next_menu = str(input("Do you wish to continue? (y/n)"))
        if (next_menu == "y"):
            continue
        else:
            break  

main()
```
