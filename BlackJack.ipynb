```python
from random import seed
from random import randrange
from datetime import datetime
from random import shuffle

#globals
kind = {"heart", "diamond", "spade", "club"}
number = {"ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "jack", "queen", "king"}
player_lost = False
player_sum = 0


deck = [(k,n) for k in kind for n in number]
players_deck = set()
computers_deck = set()


#functions
def hand_value(given_set):
    card_sum = 0
    for card in given_set:
        card_sum += card
    return card_sum

def card_value(card):
    if card in ["jack", "queen", "king"]:
        return 10
    elif card == "ace":
        ace = int(input("You want the ace equal to 1 or 11? "))
        if ace == 1:
            return 1
        else:
            return 11
    else:
        return card

def player_plays():
    global player_lost
    global player_sum
    cards_in_hand = []
    answer = "y"
    
    while answer == "y":
        print(f"--- Round {len(cards_in_hand)+1} ---")
        print("Drawing one more card. ")
        card_drawn = deck.pop()
        cards_in_hand.append(card_drawn)
        print(f"Player currently has {len(cards_in_hand)} card in his hand.")
        print(f"The card that is drawn is: {card_drawn}")
        player_sum += card_value(card_drawn[1])
        print(f"Your cards total value is: {player_sum}")
        if player_sum > 21:
            print("Exceeded 21, you lose.")
            player_lost = True
            break
        elif player_sum == 21:
            print("You hit exactly 21, you won!")
            break
        else:
            answer = input("Do you want to draw one more card? ")
            if answer == "n":
                break

    print("Player's final hand is: ",end=" -- ")
    for card in cards_in_hand:
        print (card, end = " -- ")
    print("\nPlayer's hand total value: - ||| "+str(player_sum)+" ||| -")

def computer_plays():
    print("Player sum = "+str(player_sum))
    pass
    
#main
def main():
    seed(datetime.now().timestamp())
    shuffle(deck)
    print("===== GAME STARTS =====")
    print("===== PLAYER PLAYS =====")
    player_plays()
    print(" ===== PLAYER ENDED =====")
    if (player_lost):
        computer_plays()
    
    print("\n===== GAME ENDED =====")
main()
```
