### TRAVERSE SOS

```python
string_or = "Some people never go crazy. What truly horrible lives they must lead."
```

```python
while True:
    string = input("Give word: ")
    string = string.strip()

    if string.isalpha():
        word = string.lower()
        break
    else:
        print("ae")

quote_lowered = string_or.lower()
if quote_lowered.find(word) != -1:
    print(string_or.replace(word, word.upper()))
else:
    print("Word doesn't exist.")
```
