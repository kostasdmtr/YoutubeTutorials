```python
from random import seed
from random import randrange
from datetime import datetime

#globals
kind = {"heart", "diamond", "spade", "club"}
number = {"ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "jack", "queen", "king"}

deck = {(k,n) for k in kind for n in number}

numberset = set()
numberset.add(5)
numberset.add(4)
numberset.add(3.14)
print(numberset)

#functions
def hand_value(given_set):
    card_sum = 0
    for card in given_set:
        card_sum += card
    return card_sum

def card_value(card):
    if card in ["jack", "queen", "king"]:
        return 10
    elif card == "ace":
        ace = int(input("You want the ace equal to 1 or 11? "))
        if ace == 1:
            return 1
        else:
            return 11
    else:
        return card

#main
def main():
    seed(datetime.now().timestamp)
    
    for card in number:
        print(f"card: {card} has value: {card_value(card)}")

         
main()
```
