```python
from random import seed
from random import randrange
from datetime import datetime
import random

seed(random.random())
```

```python
hidden_words = [
"fade",
"extreme",
"real",
"beneficiary",
"rock",
"smash",
"banana",
"satisfied",
"seem",
"basis",
"stab",
"stool",
"class",
"miss",
"offspring",
]

hidden_word = hidden_words[randrange(len(hidden_words))]
print(f"The hidden word we are looking for is: |{hidden_word}|")

guessed_letters = []

tries = 10

while True:
    tries -= 1
    print(f"\n==============\nTry number {10-tries}.\n=============")

    if(tries == 0):
        print("Exceeded maximum tries, you lost, sorry.")
        break
    
    letter = input("Choose a letter: ").lower()
    if (not letter.isalpha()):
        print("The thing you put was not a letter, please try again.")
        tries += 1
        continue
    elif (len(letter)>1):
        print("Provided more than one character, please try again.")
        tries += 1
        continue

    if (letter in guessed_letters):
        print("You've already picked this character. You have a 1-round penalty. Please try again.")
        tries -= 1
        continue
    guessed_letters.append(letter)
    print(f"The letter '{letter}' you chose, happens to exist {hidden_word.count(letter)} times in the hidden word.")

    print("The progress you've made so far is the following:", end = " ")
    won = True
    
    for letter in hidden_word:
        if letter in guessed_letters:
            print(letter,end ="")
        else:
            print("_",end="")
            won = False
    #ektypwsh nikhth
    if (won):
        print("\n| CONGRATULATIONS!!!, you found all the letters. |")
        print("The word was: "+hidden_word)
        break
    
```
