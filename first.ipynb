### TRAVERSE SOS

```python
from random import seed
from random import randrange
from datetime import datetime
seed(datetime.now().timestamp())

player_wins = 0
computer_wins = 0

while (player_wins<3 and computer_wins<3):

    #Eisodos Xrhsth
    player_input = input("Choose: ")
    while (player_input not in ["rock", "paper", "scissors"]):
        print("Wrong input, choose between rock, paper, scissors.")
        player_input = input("Choose: ")
    
    print(player_input)
    
    
    #Epilogh PC
    computer_random = randrange(3)
    if computer_random == 0:
        computer_choice = "rock"
    elif computer_random == 1:
        computer_choice = "paper"
    else:
        computer_choice = "scissors"
    
    print(computer_choice)
    #Elegxos poios nikhse
    
    if(player_input == "rock"):
        if computer_choice == "rock":
            winner = "no winner"
        elif computer_choice == "paper":
            winner = "computer"
        else:
            winner = "player"
    elif (player_input == "paper"):
        if computer_choice == "rock":
            winner = "player"
        elif computer_choice == "paper":
            winner = "no winner"
        else:
            winner = "computer"
    else:
        if computer_choice == "rock":
            winner = "computer"
        elif computer_choice == "paper":
            winner = "player"
        else:
            winner = "no winner"
    
    if winner == "player":
        player_wins +=1
    elif winner == "computer":
        computer_wins +=1
    
    #Ektypwsi nikiti
    print("Player: "+player_input+", Computer: "+computer_choice)
    print("Winner of this round: "+winner)
    print("Total Score: Player("+str(player_wins)+") ||| Computer("+str(computer_wins)+")")

print("player" if player_wins == 3 else "computer")
```
