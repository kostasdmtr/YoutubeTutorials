```python
friends = ["C", "Nick", "B"]
invited = ["A", "B", "C", "D", "E", "F", "G", "H"]
```

```python
friends_invited = 0
for friend in friends:
    if friend in invited:
        friends_invited+=1
        print("Your friend "+friend+" was invited!")

if friends_invited < 2:
    print("I am not coming then")
else:
    print("SO I AM COMING LETS GOOOOOOOOOO")
```
